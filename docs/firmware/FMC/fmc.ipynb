## Clock Configuration
- Direct clock supplied in clock tree
- Max frequency: 200 MHz
- Actual frequency: FMC Core clock / 2 = 100 MHz

```python
FMC_CLK = 200.0e6
CLK_DIV = 3

FMC_PERIOD = 1/FMC_CLK

SRAM_FREQ = FMC_CLK / CLK_DIV
SRAM_PERIOD = 1 / SRAM_FREQ
print(f"Period: {SRAM_PERIOD*1e9:.2f} ns")
```

# Signals

## CAS Latency
- Column address strobe latency
- Supports CAS 2 or 3 (longer to be safer)

```python
CAS_LATENCY = 3
print(f"CAS_T: {CAS_LATENCY * SRAM_PERIOD*1e9 : .2f} ns")
```

## Read Pipe Delay
Compensates for SDRAM access time, propagation and GPIO delays etc..

```python
READ_PIPE_DELAY = 3
print(f"READ_PIPE_DELAY_T: {READ_PIPE_DELAY * FMC_PERIOD*1e9 : .2f} ns")
```

```python
tRSC = 2 * FMC_PERIOD
print(f"MODE REGISTER DELAY: {tRSC*1e9:.2f} ns")
```

## Exit self-refresh to ACTIVE command

```python
import math
txSR = 72e-9
EXIST_SLEF_REFRESH_TO_ACTIVE = txSR * SRAM_FREQ
print(f"CYCLES: {math.ceil(EXIST_SLEF_REFRESH_TO_ACTIVE)}")
```
